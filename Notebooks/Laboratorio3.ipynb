```python
# IMPORTAMOS LIBRERÍAS
import pandas as pd
import random
```

```python
# DATOS BASE
datosInstagram = {
    "influencer": [
        "ana_travel",
        "luis_beach",
        "cancun_life",
        "mexico_vibes",
        "playas_mx",
        "travel_queen"
    ],

    "tipo_post": [
        "turismo",
        "playa",
        "lifestyle",
        "comida",
        "hotel",
        "aventura"
    ],

    "fecha": [
        "2026-01-10",
        "2026-02-14",
        "2026-03-20",
        "2026-04-15",
        "2026-05-18",
        "2026-06-01"
    ]
}
```

```python
# CREAR DATAFRAME BASE
dfInstagram = pd.DataFrame(datosInstagram)
```

```python
# MOSTRAR DATOS BASE
dfInstagram
```

```python
# CANTIDAD DE REGISTROS
cantidad = 1000000
```

```python
# GENERAR BIG DATASET
dfBigData = pd.DataFrame({

    "influencer": [
        random.choice(dfInstagram["influencer"])
        for _ in range(cantidad)
    ],

    "likes": [
        random.randint(100, 50000)
        for _ in range(cantidad)
    ],

    "comentarios": [
        random.randint(10, 5000)
        for _ in range(cantidad)
    ],

    "fecha": [
        random.choice(dfInstagram["fecha"])
        for _ in range(cantidad)
    ],

    "tipo_post": [
        random.choice(dfInstagram["tipo_post"])
        for _ in range(cantidad)
    ]

})
```

```python
# MOSTRAR LAS PRIMERAS FILAS
dfBigData.head()
```

```python
# TAMAÑO DEL DATAFRAME
dfBigData.shape
```

```python
# TIPOS DE DATOS
dfBigData.dtypes
```

```python
# MEMORIA UTILIZADA
dfBigData.memory_usage(deep=True).sum() / 1024**2
```

```python
dfBigData["likes"].mean()
```

```python
# Influencer con más likes promedio
dfBigData.groupby("influencer")["likes"].mean()
```

```python
#Tipo de post con más comentarios
dfBigData.groupby("tipo_post")["comentarios"].mean()
```

```python
#Total de publicaciones por tipo
dfBigData["tipo_post"].value_counts()
```

```python
#Top 10 publicaciones con más likes
dfBigData.sort_values(by="likes", ascending=False).head(10)
```
